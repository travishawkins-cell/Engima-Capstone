## Text Processing V5

```python
import numpy as np
import unicodedata

# Dictionary to map digits to their spelled-out forms
digit_map = {
    '0': 'zero', '1': 'one', '2': 'two', '3': 'three', '4': 'four',
    '5': 'five', '6': 'six', '7': 'seven', '8': 'eight', '9': 'nine'
}

# Load the text file
input_filename = 'The home book of verse.txt'
with open(input_filename, 'r', encoding='utf-8') as file:
    text = file.read()

# Process in one pass: replace digits, remove accents, keep only plain letters
result_chars = []
for char in text:
    if char.isdigit():
        # Add spelled-out digit (already lowercase)
        result_chars.extend(digit_map[char])
    elif char.isalpha():
        # Convert to lowercase
        char_lower = char.lower()
        # Remove accents 
        char_normalized = unicodedata.normalize('NFD', char_lower)
        char_plain = ''.join(c for c in char_normalized if unicodedata.category(c) != 'Mn')
        # Only add if it's a plain ASCII letter
        if char_plain.isalpha():
            result_chars.append(char_plain)

# Convert to numpy matrix
matrix = np.array(result_chars).reshape(-1, 1)

# Display the matrix
print("Matrix shape:", matrix.shape)
print("\nFirst 20 rows of the matrix:")
print(matrix[:20])

# Generate output filename
base_name = input_filename.rsplit('.', 1)[0]
extension = input_filename.rsplit('.', 1)[1] if '.' in input_filename else 'txt'
output_filename = f"{base_name} processed.{extension}"

# Save the matrix
np.savetxt(output_filename, matrix, fmt='%s', delimiter='')
print(f"\nOutput saved to: {output_filename}")
```

## Text processing V4

```python
import numpy as np

# Dictionary to map digits to their spelled-out forms
digit_map = {
    '0': 'zero', '1': 'one', '2': 'two', '3': 'three', '4': 'four',
    '5': 'five', '6': 'six', '7': 'seven', '8': 'eight', '9': 'nine'
}

# Load the text file
input_filename = 'The home book of verse.txt'
with open(input_filename, 'r') as file:
    text = file.read()

a = 0
b = len(text)

# Replace all digit characters with their spelled-out forms
processed_chars = []  # Use list instead of string
for char in text:
    a = a + 1
    if char.isdigit():
        processed_chars.append(digit_map[char])
    else:
        processed_chars.append(char)
    if a % 1000000 == 0:
        print(a)

processed_text = ''.join(processed_chars)  # Join once at the end

# Remove punctuation and spaces - keep only alphanumeric characters
cleaned_text = ''.join(char for char in processed_text if char.isalnum())

# Create a list of all characters
char_list = list(cleaned_text)

# Convert to numpy matrix with 1 column
matrix = np.array(char_list).reshape(-1, 1)

# Display the matrix
print("Matrix shape:", matrix.shape)
print("\nFirst 20 rows of the matrix:")
print(matrix[:20])

# Generate output filename
base_name = input_filename.rsplit('.', 1)[0]
extension = input_filename.rsplit('.', 1)[1] if '.' in input_filename else 'txt'
output_filename = f"{base_name} processed.{extension}"

# Save the matrix
np.savetxt(output_filename, matrix, fmt='%s', delimiter='')
print(f"\nOutput saved to: {output_filename}")
```
